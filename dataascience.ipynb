```python
import matplotlib.pyplot as plt
import numpy as np

# Data from Oxford Insights Government AI Readiness Index 2024
categories = ["Overall", "Government", "Technology\nSector", "Data &\nInfrastructure"]

kazakhstan = [51.41, 54.75, 33.54, 65.93]
uzbekistan = [53.45, 64.71, 33.50, 62.14]

x = np.arange(len(categories))
width = 0.36

plt.figure(figsize=(7, 4.5))

bars1 = plt.bar(
    x - width/2,
    kazakhstan,
    width,
    label="Kazakhstan",
    color="#4C78A8",      # soft blue
    edgecolor="black"
)

bars2 = plt.bar(
    x + width/2,
    uzbekistan,
    width,
    label="Uzbekistan",
    color="#F58518",      # soft orange
    edgecolor="black"
)

plt.ylabel("Score")
plt.ylim(0, 80)
plt.xticks(x, categories)
plt.title("Government AI Readiness Index 2024")
plt.legend(frameon=False)
plt.grid(axis="y", linestyle="--", linewidth=0.5, alpha=0.6)

# Add values above bars
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + 1,
            f"{height:.2f}",
            ha="center",
            va="bottom",
            fontsize=8
        )

plt.tight_layout()
plt.savefig("figure_1_ai_readiness_kazakhstan_uzbekistan.png", dpi=300, bbox_inches="tight")
plt.show()
```
